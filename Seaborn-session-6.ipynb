```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Swiggy data
data = {
    "delivery_time": [
        20, 25, 30, 35, 40, 45, 25, 30, 35, 40,
        50, 55, 30, 35, 45, 50, 25, 40, 45, 60,
        30, 35, 40, 50, 55, 65, 25, 30, 45, 50
    ],
    
    "order_value": [
        200, 250, 300, 350, 400, 450, 220, 280, 330, 380,
        500, 550, 310, 360, 420, 480, 230, 390, 440, 600,
        320, 370, 410, 510, 560, 650, 240, 290, 450, 520
    ]
}

df = pd.DataFrame(data)

# Hexbin jointplot
sns.jointplot(
    x="delivery_time",
    y="order_value",
    data=df,
    kind="hex"
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "rating": [
        2.5, 3.0, 3.2, 3.5, 3.8, 4.0, 4.1, 4.2, 4.3, 4.5,
        4.6, 4.7, 4.8, 4.0, 3.9, 3.5, 4.2, 4.4, 4.6, 4.8,
        3.0, 3.3, 3.7, 4.1, 4.5, 4.7, 3.8, 4.3, 4.6, 4.9
    ],
    
    "number_of_reviews": [
        20, 30, 35, 40, 50, 60, 70, 80, 90, 100,
        110, 120, 130, 65, 55, 45, 75, 95, 115, 140,
        25, 38, 48, 72, 105, 125, 58, 88, 118, 150
    ]
}

df = pd.DataFrame(data)
```

```python
# 4 A scatter plot
```

```python
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=df,
    kind="scatter"
)

plt.show()
```

```python
# 4 B Hexbin plot
```

```python
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=df,
    kind="hex"
)

plt.show()
```

```python

# 4c KDE plot
```

```python
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=df,
    kind="kde"
)

plt.show()
```
